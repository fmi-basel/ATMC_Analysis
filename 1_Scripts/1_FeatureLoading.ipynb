# User Input

```python
source = "YOUR PATH TO EXPERIMENT FOLDER" # Point to experiment folder. Has to include setup .xls file.

analysis_dir = None # Optional: Analysis directory. If None this will default to the source folder. Creates 4 folder substructure for standardization in subsequent notebooks.

file_ending = ".zarr" # File ending to search for ome_zarr files.

experiment_ID = "YOUR EXPERIMENT ID"

feature_table_names = "YOUR_FEATURE_TABLE_NAME" # Feature table(s) to load (str or list[str])

# Multiplexing round(s) to load from .zarr/<row>/<col>/<round>/...
# Either int (single round) or list[int] (multiple rounds).
# No multiplexing: set to 0
multiplexing_rounds = 0

roi_table_name = "nuclei_ROI_table" # Fractal ROI table name to load images.
label_name = "nuclei" # Fractal label name to load segmentation.
```

# Load files and display experimental setup

```python
# ── Run Cell ───────────────────────────────────────────────────────────────────────
stainings = get_stainings(source)
```

```python
# ── Run Cell ───────────────────────────────────────────────────────────────────────
folder, analysis_dir = find_zarr_dirs(source, file_ending = file_ending, analysis_dir = analysis_dir) # Find ome_zarr files in source folder.
experiment_setup, barcodes = make_experiment(source) # Create experiment setup from Layout in source folder.
```

# Load + merge features (and save)

```python
# ── Run Cell ───────────────────────────────────────────────────────────────────────
ad = merge_feature_tables_from_zarr_rounds(
    source = source,
    folder = folder,
    experiment_setup = experiment_setup,
    stainings = stainings,
    experiment_ID = experiment_ID,
    multiplexing_rounds = multiplexing_rounds,
    result_file_name = "1_FeatureLoading",
    analysis_dir = analysis_dir,
    feature_table_names = feature_table_names,
    roi_table_name = roi_table_name,
    label_name = label_name,
    file_ending = file_ending,
)

print('Merged AnnData ready:')
print('  Objects:', ad.n_obs)
print('  Features:', ad.n_vars)
print('  Multiplexing Rounds:', ad.uns.get('multiplexing_rounds', multiplexing_rounds))
```
